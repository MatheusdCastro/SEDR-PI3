```python
import numpy as np
from scipy.io import wavfile

def audio_to_binary_txt(wav_filename, txt_filename):
    sample_rate, data = wavfile.read(wav_filename)

    if len(data.shape) > 1:
        data = data[:, 0]

    # Normaliza o áudio para o intervalo [-1.0, 1.0]
    if data.dtype == np.int16:
        data_norm = data / 32768.0
    elif data.dtype == np.int32:
        data_norm = data / 2147483648.0
    else:
        data_norm = data / np.max(np.abs(data))

    # formato de 16 bits
    fixed_point_data = np.round(data_norm * 32767).astype(np.int64)

    with open(txt_filename, 'w') as f:
        for sample in fixed_point_data:
            # Complemento de 2
            if sample < 0:
                sample = (1 << 16) + sample
            binary_str = format(sample & 0xFFFF, '016b')
            f.write(binary_str + '\n')

    print(f"Sucesso! Arquivo '{txt_filename}' gerado com {len(fixed_point_data)} amostras.")
    print(f"Taxa de amostragem original: {sample_rate} Hz")


audio_to_binary_txt('sample-6s.wav', 'entrada_audio_16bits.txt')
```
